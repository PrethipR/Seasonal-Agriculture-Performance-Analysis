1. PROBLEM STATEMENT

Agricultural performance varies across seasons due to environmental conditions, farming practices, resource availability and market conditions. The objective is to analyze the agricultural dataset and identify seasonal differences, trends, relationships and variations in agricultural performance.

2. PROJECT OBJECTIVES

-To analyze agricultural performance across different seasons using data analytics techniques.
-To identify trends and patterns in crop yield, production, revenue, cost, and profit.
-To compare the performance of different crops, states, and irrigation methods.
-To study the impact of environmental factors such as rainfall, temperature, humidity, and soil moisture on crop
yield.
-To analyze resource usage such as water, fertilizer, pesticides, and seed quality.
-To identify disease and pest risk patterns across crops and seasons.
-To measure water efficiency and identify better resource-management practices.
-To provide data-driven recommendations that can help improve agricultural productivity, profitability, and resource efficiency.

```python
#import the libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

print("Libraries imported successfully")
```

Load the Dataset

```python
file_path = "seasonal_agriculture_performance_dataset (2).csv"

df = pd.read_csv(file_path)

print("Datasetloaded successfully")
print("Number of Rows:", df.shape[0])
print("NUmber of Columns:", df.shape[1])

#dataset information
df.info()
```

Check missing values

```python
missing_values = df.isnull().sum()

missing_values = missing_values[missing_values > 0]

print("Missing Values:")
print(missing_values)
```

Missing value Percentage

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage = missing_percentage[missing_percentage > 0]

print(missing_percentage.round(2))
```

Handle Missing Values

```python
numeric_columns = [
    "Rainfall_mm",
    "Soil_Moisture_pct",
    "Yield_Tonnes_Ha"
]

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("Missing values handled successfully.")
```

```python
#check again for missing values
print(df.isnull().sum().sum())
```

Check Duplicate values

```python
duplicates = df.duplicated().sum()

print("Number of duplicate records:", duplicates)

#Remove the dublicate values
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```
